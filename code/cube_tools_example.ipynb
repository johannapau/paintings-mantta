```python
import numpy as np
import matplotlib.pyplot as plt
import os


camera = 'IQ'
path = f'../data/{camera}/reflectance' 
os.listdir(path)
```

```python
painting = '003c'
cube = np.load(f'{path}/{painting}.npy')
cube.ndim
```

# CubeClipper

* Snipping Tool -like HS cube clipper --> Click and drag to outline an area in the image to be able to easily save smaller subimages 

```python
from cube_tools import CubeClipper
%matplotlib widget

band = 60  # The band you want displayed
savepath = 'clippings'  # path to folder where the clips are saved. 
clipper = CubeClipper(cube, band=60, out_dir=savepath)
```

# Spectral Explorer

* Click or lasso areas of the HS cube and plot (average) spectra.

```python
from cube_tools import SpectralExplorer
%matplotlib widget

wls = range(cube.shape[-1])  # Wavelengths here!
explorer = SpectralExplorer(cube, wavelengths=wls, rgb_bands=(60, 30, 10))
```

# BandSurfer

```python
from cube_tools import BandSurfer
%matplotlib widget

wls = range(cube.shape[-1])  # Wavelengths here!

# --- Usage --- 
surfer = BandSurfer(cube, wavelengths=wls)
```
